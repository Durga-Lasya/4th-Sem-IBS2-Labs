```python
sequence="""MEEPQSDPSVEPPLSQETFSDLWKLLPENNVLSPLPSQAMDDLMLSPDDIEQWFTEDPGP
DEAPRMPEAAPPVAPAPAAPTPAAPAPAPSWPLSSSVPSQKTYQGSYGFRLGFLHSGTAK""".replace("\n","")

hydrophobic={'A','V','I','L','M','F','Y','W'}

def score(seq):
    return sum(2 if r in hydrophobic else 1 for r in seq)

def expand_consecutive(candidates,sequence):
    new=[]
    for cand in candidates:
        if cand=="":
            for i in range(len(sequence)):
                new.append(sequence[i])
        else:
            for i in range(len(sequence)-len(cand)):
                if sequence[i:i+len(cand)]==cand:
                    new.append(sequence[i:i+len(cand)+1])
    return list(set(new))

def trim(leaderboard,N):
    scored=[(p,score(p)) for p in leaderboard]
    scored.sort(key=lambda x:x[1],reverse=True)
    print("\nSorted Candidates:")
    for p,s in scored:
        print(f"{p} Score={s}")
    if len(scored)<=N:
        return [p for p,s in scored]
    cutoff=scored[N-1][1]
    print(f"\nCutoff Score (Top {N})={cutoff}")
    trimmed=[p for p,s in scored if s>=cutoff]
    print("\nAfter Trimming:")
    for p in trimmed:
        print(f"{p} Score={score(p)}")
    return trimmed

def leaderboard(sequence,k=5,N=5):
    leaderboard=[""]
    step=1
    while step<=k:
        print("\n==============================")
        print(f"STEP {step} (Building {step}-mers)")
        print("==============================")
        leaderboard=expand_consecutive(leaderboard,sequence)
        print(f"\nAfter Expansion ({len(leaderboard)} candidates):")
        print(leaderboard)
        leaderboard=trim(leaderboard,N)
        step+=1
    print("\nFINAL 5-MERS:")
    for p in leaderboard:
        print(f"{p} Score={score(p)}")
    return max(leaderboard,key=score)

best_peptide=leaderboard(sequence,k=5,N=5)
print("\nFinal Best 5-mer:",best_peptide)
print("Final Score:",score(best_peptide))
```
